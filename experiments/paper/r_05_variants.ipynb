```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array(
    [
        [0, 0.0008, 0.054, 0.12, 0.072, 0.22, 0.62],
        [0.001, 0, 0.050, 0.11, 0.081, 0.23, 0.68],
        [0.021, 0.016, 0, 0.12, 0.027, 0.22, 0.63],
        [0.12, 0.097, 0.13, 0, 0.0094, 0.25, 0.71],
        [0.085, 0.082, 0.087, 0.028, 0, 0.27, 0.78],
        [0.16, 0.16, 0.17, 0.25, 0.23, 0, 0.55],
        [0.63, 0.69, 0.64, 0.63, 0.67, 0.53, 0],
    ]
)

A = 1 / 2 * (A + A.T)

# Create figure and axis
fig, ax = plt.subplots(figsize=(8, 15))

# Display the matrix
im = ax.imshow(A)

# Axis labels for both dimensions
labels = ["20x10", "Subd.", "10x10", "30x30", "40x40", "Irreg. theta", "Irreg. phi"]
ax.set_xticks(np.arange(len(labels)))
# ax.set_yticks(np.arange(len(labels)))
# ax.set_xticklabels(labels, rotation=30, ha="right")
ax.set_xticklabels(labels, fontsize=10)
# ax.set_yticklabels(labels, fontsize=10)
# hide yticks
ax.set_yticks([])

# Add text labels in the center of each block
vmin = A.min()
vmax = A.max()
for i in range(A.shape[0]):
    for j in range(A.shape[1]):
        value = A[i, j]
        # Choose text color based on background brightness
        text_color = "white"
        ax.text(
            j,
            i,
            f"{value:.3f}",
            ha="center",
            va="center",
            color=text_color,
            fontsize=10,
            weight="bold",
        )

# Add colorbar to the right of the matrix
# plt.tight_layout()

# Get the position of the matrix axes after tight_layout
pos = ax.get_position()

# Create colorbar axes to the right of the matrix
cbar_width = 0.02
cbar_pad = 0.02
cbar_ax = fig.add_axes([pos.x1 + cbar_pad, pos.y0, cbar_width, pos.height])
cbar = plt.colorbar(im, cax=cbar_ax, orientation="vertical")
plt.savefig("distance_matrix_tori_05.png", dpi=300, bbox_inches="tight")
plt.show()
```
