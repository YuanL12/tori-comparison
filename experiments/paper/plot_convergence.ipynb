```python
import matplotlib.pyplot as plt
import numpy as np
import os
import pandas as pd
import seaborn as sns
import sys

# Example: load CSV and plot Step vs. b2_vs_e2 - energy

# Path to your CSV file
csv_path = "e2_b2_energy_result.csv"  # update with your filename if needed

# Load the CSV file
df = pd.read_csv(csv_path)
```

```python
df.head(5)
```

```python

plt.figure(figsize=(10,6))
plt.plot(df["Step"], df["e2_vs_b2(geo) - energy"], label="r_09_vs_r_05")
# plt.plot(df["Step"], df["e2_vs_b2(euc) - energy"], label="e2_vs_b2(euc) - energy")
plt.plot(df["Step"], df["b2_vs_e2(geo) - energy"], label="r_05_vs_r_09")
# plt.plot(df["Step"], df["b2_vs_e2(euc) - energy"], label="b2_vs_e2(euc) - energy")
plt.xlabel("Step")
plt.ylabel("Energy")
# plt.title("Convergence Plot: Energy vs Step")
plt.xlim(right=33000, left=0)
plt.legend()
plt.grid(True)
plt.savefig("e2_b2_energy_result.pdf", bbox_inches="tight", dpi=300)
plt.show()
```
